```python
import pandas as pd

df = pd.DataFrame(
    cancer.data,
    columns=cancer.feature_names
)

df
```

```python
df['target_original'] = cancer.target

df['target'] = 1 - df['target_original']

df['target_name'] = df['target'].map({
    0:'benign', 1:'malignant'
})

df.head()
```

```python
df[['target_original', 'target', 'target_name']].drop_duplicates()
```

```python
data = load_breast_cancer()

feature_names = list(data.feature_names)
```

```python
X = pd.DataFrame(data.data, columns=feature_names)

X
```

```python
y_original = pd.Series(data.target, name='target_original')

y_original
```

```python
y = 1 - y_original

y
```

```python
y.name = 'target'

y
```

```python
y.map({0:'양성', 1:'악성'}).value_counts()
```

```python
label_to_name = {0:'benign', 1:'malignant'}

target_name = y.map(label_to_name)

target_name.name = 'target_name'

target_name
```

```python
df = X.copy()

df
```

```python
df['target'] = y.values

df
```

```python
df['target_name'] = target_name.values

df
```

```python
count_by_class = df['target_name'].value_counts()

count_by_class
```

```python
count_by_class.index
```

```python
count_by_class.values
```

```python
import matplotlib.pyplot as plt

plt.bar(count_by_class.index, count_by_class.values)
plt.title('Target Class Distribution')
plt.xlabel('Class')
plt.ylabel('Count')

for i, value in enumerate(count_by_class.values):
    plt.text(i, value+3, str(value), ha='center')
    
plt.show()
```

```python
mean_compare = df.groupby('target_name')[feature_names].mean().T
```

```python
mean_compare['difference'] = mean_compare['malignant'] - mean_compare['benign'] # 양수: 악성에서의 평균이 더 큼
```

```python
mean_compare['abs_difference'] = mean_compare['difference'].abs()
```

```python
mean_compare.sort_values('abs_difference', ascending=False).head(15)
```
