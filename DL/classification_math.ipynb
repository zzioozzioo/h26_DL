```python
# 파이썬에서 수학적인 연산을 쉽게 하도록 제공하는 math 라이브러리에서
# 로그(logarithm)를 계산할 수 있는 log 함수를 가져옵니다.
from math import log
```

```python
# log(x, base) 함수 설명:
# 이 함수는 'base(밑)'를 몇 번 곱해야 'x'가 되는지를 알려주는 로그 값을 계산합니다.
# 즉, log(base)를 기준으로 값(x)의 로그 값을 구하는 함수입니다.
# 예를 들어, log(8, 2)는 '2를 몇 번 곱하면 8이 되는지'를 계산합니다.
```

```python
# 예시
# 밑(base)이 2인 로그에서 숫자 8의 로그 값을 계산합니다.
log_value = log(8, 2)

# 위의 계산 결과를 출력하여 확인합니다.
print(log_value)
```

```python
# 예시
# 밑(base)이 10인 로그에서 숫자 1000의 로그 값을 계산합니다.
log_value = log(1000, 10)

# 위의 계산 결과를 출력하여 확인합니다.
print(log_value)
```
