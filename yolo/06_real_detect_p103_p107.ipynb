```python
from ultralytics import YOLO
```

```python
model = YOLO(r"D:\Lecture\ai_project01\yolo\results\mask_detection\weights\best.pt")
```

```python
model.predict(
    source=r"D:\Lecture\ai_project01\yolo\mask_yolo_dataset\images\val",
    save=True,
    project=r"D:\Lecture\ai_project01\yolo\runs\detect",
    name="predict",
    exist_ok=True
)
```

```python
import cv2
cap = cv2.VideoCapture(0)
```

```python
print("실시간 마스크 탐지를 시작합니다. 'q'를 누르면 종료됩니다.")

while True:
    if not cap.isOpened():
        print("웹캠을 열 수 없습니다.")
        break

    ret,frame = cap.read()

    if not ret:
        print("프레임을 읽을 수 없습니다.")
        break

    results = model.predict(source=frame, conf=0.5, imgsz=640, verbose=False)

    annotated_frame = results[0].plot()

    cv2.imshow("Mask Detection(YOLO plot 사용)", annotated_frame)

    if cv2.waitKey(1) == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
